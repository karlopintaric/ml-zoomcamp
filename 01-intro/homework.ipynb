```python
median_val = df.horsepower.median()
mode_val = df.horsepower.mode()
df["horsepower"] = df.horsepower.fillna(mode_val)
new_median_val = df.horsepower.median()

print(f"Old median: {median_val}, new median: {new_median_val}")
```

```python
df_asia_sample = df.loc[df.origin == "Asia", ["vehicle_weight", "model_year"]].head(7)
X = df_asia_sample.to_numpy()
X
```

```python
XTX = X.T@X
```

```python
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = XTX_inv@X.T@y
```
